```python
def outlier_capping(df,column):
    Q1=df[column].quantile(0.25)
    Q3=df[column].quantile(0.75)
    IQR=Q3-Q1
    lower_extreme=Q1-1.5*IQR
    upper_extreme=Q3+1.5*IQR
    df[column]=df[column].apply(lambda x:lower_extreme if x<lower_extreme else upper_extreme if x>upper_extreme else x)
for col in df.select_dtypes(['int','float']).columns:
    outlier_capping(df,col)
```

## Interview Question

### 1. Difference between precision and recall

Precision tells you, out of all the items your model predicted as positive, how many were actually positive. It reflects how reliable the positive predictions are. High precision means the model rarely gives a positive label when it should not.

Recall tells you, out of all the actual positive cases, how many your model correctly identified. It focuses on not missing real positives. High recall means the model catches most of the true positive cases, even if it sometimes predicts positive when it should not.

In simple terms: precision cares about being accurate when saying “yes,” while recall cares about catching all the true “yes” cases.

### 2. Cross-validation and its importance in binary classification

Cross-validation is a technique where the dataset is split into multiple parts, and the model trains on some of them while being tested on the remaining part. This process rotates so every portion of data gets a chance to be used for testing. Instead of judging the model based on just one train-test split, cross-validation gives a more reliable picture of how it performs on unseen data.

In binary classification, cross-validation matters because it helps prevent overfitting and gives a fair estimate of the model’s ability to separate the two classes. It ensures that performance metrics like accuracy, precision, recall, or F1-score are trustworthy and not just the result of a lucky data split.
